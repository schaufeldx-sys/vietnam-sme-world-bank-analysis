## 3. Strategic Recommendations

### For Commercial Banks Entering/Expanding in Vietnam:

---

#### RECOMMENDATION 1: Target the Underserved Small Firm Segment

| Aspect | Details |
|--------|---------|
| **Target** | Small firms (5-19 employees) in manufacturing and services |
| **Gap** | 42.5% have bank accounts, 28.4% have loans |
| **Action** | Develop tailored products for this segment |
| **Potential** | ~620 firms in sample = significant market |

---

#### RECOMMENDATION 2: Offer Collateral Alternatives

| Aspect | Details |
|--------|---------|
| **Problem** | 93.5% of loans require collateral (vs 67% globally) |
| **Barrier** | SMEs often lack traditional collateral (real estate) |
| **Solutions** | Accept: equipment, inventory, receivables, group guarantees |
| **Benefit** | Unlocks credit for firms currently excluded |

---

#### RECOMMENDATION 3: Expand Financial Inclusion Infrastructure

| Aspect | Details |
|--------|---------|
| **Problem** | 45% of firms lack bank accounts |
| **Opportunity** | Basic accounts lead to deeper relationships |
| **Action** | Partner with mobile money, simplify KYC |
| **Result** | Pipeline for future loan products |

---

#### RECOMMENDATION 4: Capture Female Entrepreneur Market

| Aspect | Details |
|--------|---------|
| **Segment** | Firms with female top managers (21.5% of sample) |
| **Gap** | 13 pp less likely to have bank accounts than male-owned |
| **Action** | Targeted products and outreach programs |
| **Benefit** | Less competition, loyal customers |

---

#### RECOMMENDATION 5: Simplify Lending Processes

| Aspect | Details |
|--------|---------|
| **Problem** | Perceived complexity deters applications |
| **Vietnam Rate** | Only 1.9% rejection (lower than global) |
| **Interpretation** | Qualified firms exist but don't apply |
| **Action** | Clear communication, fast decisions, digital apply |


## 4. Implementation Roadmap

```python
# Create timeline visualization
fig, ax = plt.subplots(figsize=(14, 6))

# Timeline
phases = [
    ('Phase 1\n(Month 1-3)', 0.15, 'Foundation', '#457B9D', [
        '• Market research deep-dive',
        '• Product design workshops',
        '• Partner identification',
    ]),
    ('Phase 2\n(Month 4-6)', 0.45, 'Development', '#2A9D8F', [
        '• Build collateral alternatives',
        '• Create SME onboarding flow',
        '• Staff training program',
    ]),
    ('Phase 3\n(Month 7-12)', 0.75, 'Launch', '#E63946', [
        '• Pilot with 50 firms',
        '• Iterate based on feedback',
        '• Scale to 200+ firms',
    ]),
]

for phase_name, x, title, color, items in phases:
    # Draw circle
    circle = plt.Circle((x, 0.5), 0.12, color=color, transform=ax.transAxes)
    ax.add_patch(circle)
    
    # Phase name
    ax.text(x, 0.5, phase_name, ha='center', va='center', fontsize=9, 
            color='white', fontweight='bold', transform=ax.transAxes)
    
    # Title above
    ax.text(x, 0.78, title, ha='center', va='bottom', fontsize=14, 
            fontweight='bold', transform=ax.transAxes)
    
    # Items below
    for i, item in enumerate(items):
        ax.text(x, 0.28 - i*0.08, item, ha='center', va='top', fontsize=10, 
                transform=ax.transAxes)

# Draw connecting line
ax.plot([0.03, 0.97], [0.5, 0.5], 'k-', linewidth=2, transform=ax.transAxes)

ax.text(0.5, 0.9, 'IMPLEMENTATION ROADMAP', ha='center', va='bottom', 
        fontsize=18, fontweight='bold', transform=ax.transAxes)

ax.axis('off')

plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'implementation_roadmap.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Saved: {OUTPUT_DIR / 'implementation_roadmap.png'}")
```
