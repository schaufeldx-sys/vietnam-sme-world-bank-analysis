## 4. Key Findings Summary

```python
# Extract key statistics for summary
summary = {
    'Metric': [
        'Firms with credit constraints (%)',
        'Firms with bank account (%)',
        'Firms with bank loan (%)',
        'Loans requiring collateral (%)',
        'Loan applications rejected (%)',
        'Investment from internal funds (%)',
        'Investment from banks (%)',
    ],
    'Vietnam': [26.2, 54.8, 40.5, 93.5, 1.9, 68.4, 15.7],
    'East Asia & Pacific': [20.9, 85.5, 34.8, 64.4, 3.0, 76.9, 13.6],
    'All Economies': [31.0, 89.9, 36.3, 67.1, 6.5, 72.7, 14.2]
}

summary_df = pd.DataFrame(summary)
print("=" * 70)
print("KEY INDICATORS SUMMARY - VIETNAM 2023")
print("=" * 70)
display(summary_df)
```

## 5. Observations

### Key Observations from Indicator Data:

1. **Credit Constraints**: 26.2% of Vietnamese firms face financing constraints, slightly above the East Asia average (20.9%) but below the global average (31.0%).

2. **Bank Account Penetration**: Only 54.8% of firms have a bank account - significantly lower than regional (85.5%) and global (89.9%) averages. This suggests a large informal finance sector.

3. **Collateral Requirements**: 93.5% of loans require collateral - much higher than regional (64.4%) and global (67.1%) averages. This is a major barrier for SMEs.

4. **Loan Rejection Rate**: Only 1.9% of applications are rejected - lower than regional and global averages, but this may reflect selection bias (only creditworthy firms apply).

5. **Financing Sources**: Vietnamese firms rely heavily on internal funds (68.4%) for investment, slightly below regional average (76.9%).

### Data Limitations:
- These are aggregate indicators, not firm-level data
- Cannot cross-tabulate by firm characteristics (size, industry, ownership)
- Pending access to firm-level survey data for deeper analysis
